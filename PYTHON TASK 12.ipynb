```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
from sklearn.model_selection import train_test_split
```

```python
df=pd.read_csv("business_sales.csv")
print("first 5 records:")
print(df.head())

print("dataset shape:")
print(df.shape)
```

```python
print("missing values:")
print(df.isnull().sum())
```

```python
x = df[["Advertising_Expense", "Discount_Percent", "Customers"]]
y = df["Sales"]
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
y_prend=model.predict(x_test)
y_prend
```

```python
result = pd.DataFrame({"Actual_sales": y_test, "Predicted_sales": y_prend})
print("\nActual_sales vs Predicted_sales")
print(result.head(10))
```

```python
mse = mean_squared_error(y_test, y_prend)
r2 = r2_score(y_test, y_prend)
print("M. S. E : ", mse)
print("R2 score : ", r2)
```
