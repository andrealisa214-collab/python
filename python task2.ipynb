```python
transactions = [500, -200, 1000, -700]

for amount in transactions:
    if amount > 0:
        print("Credit")
    else:
        print("Debit")
```

```python
transactions = [2000, 6000, 4500, 8000, 3000]

for amount in transactions:
    if amount > 5000:
        print("Large")
    else:
        print("Small")
```

```python
balance = 10000

amount = int(input("Enter debit amount: "))

if amount <= balance:
    print("Debit successful")
else:
    print("Insufficient balance")
```

```python
account = {"balance": 8000}

amount = int(input("Enter amount: "))

if amount <= account["balance"]:
    account["balance"] = account["balance"] - amount
    print("New balance:", account["balance"])
else:
    print("Not enough balance")
```

```python
transactions = [("Credit", 1000), ("Debit", 500)]

for transaction in transactions:
    if transaction[0] == "Credit":
        print("Added")
    else:
        print("Deducted")
```
